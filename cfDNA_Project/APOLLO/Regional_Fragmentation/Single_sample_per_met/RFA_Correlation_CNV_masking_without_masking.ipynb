```python
import pandas as pd

MASKED = "/ix1/alee/LO_LAB/Personal/Rahul/cfDNA_project/Broad/NST/Fragmentation_analysis/Regional_Fragment_analysis/LOLA_inputs_top3sites/Effect_size/KW_BH/KW_BH.universe50.rawKWp_sig.with_site_direction.with_genes.tsv"
UNMASKED = "/ix1/alee/LO_LAB/Personal/Rahul/cfDNA_project/Broad/NST/Fragmentation_analysis/Regional_Fragment_analysis/LOLA_inputs_top3sites/Effect_size_no_cnv_masking/KW_BH/KW_BH.no_cnv_masking.rawsig.with_site_direction.with_genes.tsv"

# -------------------------
# Load already-filtered RAW-significant files
# -------------------------
df_mask = pd.read_csv(MASKED, sep="\t", dtype={"bin_id": str})
df_unmask = pd.read_csv(UNMASKED, sep="\t", dtype={"bin_id": str})

print("Masked raw-significant bins   :", df_mask.shape[0])
print("Unmasked raw-significant bins :", df_unmask.shape[0])

# keep unique bins only
df_mask = df_mask.drop_duplicates(subset=["bin_id"]).copy()
df_unmask = df_unmask.drop_duplicates(subset=["bin_id"]).copy()

# -------------------------
# Bin-set comparison
# -------------------------
mask_bins = set(df_mask["bin_id"])
unmask_bins = set(df_unmask["bin_id"])

shared_bins = mask_bins & unmask_bins
mask_only   = mask_bins - unmask_bins
unmask_only = unmask_bins - mask_bins

print("\nBin overlap summary:")
print("Masked total bins    :", len(mask_bins))
print("Unmasked total bins  :", len(unmask_bins))
print("Shared bins          :", len(shared_bins))
print("Masked-only bins     :", len(mask_only))
print("Unmasked-only bins   :", len(unmask_only))

# -------------------------
# site_highest counts
# -------------------------
mask_counts = df_mask["site_highest"].value_counts().rename("masked")
unmask_counts = df_unmask["site_highest"].value_counts().rename("unmasked")

comparison = pd.concat([mask_counts, unmask_counts], axis=1).fillna(0).astype(int)
comparison["difference_unmasked_minus_masked"] = comparison["unmasked"] - comparison["masked"]

print("\nMasked site_highest counts:")
print(mask_counts)

print("\nUnmasked site_highest counts:")
print(unmask_counts)

print("\nComparison table:")
print(comparison)
```

```python
import os
import matplotlib.pyplot as plt

OUTDIR = "/ix1/alee/LO_LAB/Personal/Rahul/cfDNA_project/Broad/NST/Fragmentation_analysis/Regional_Fragment_analysis/LOLA_inputs_top3sites/Comparison_CNV_noCNV/"
os.makedirs(OUTDIR, exist_ok=True)

plot_df = comparison[["masked", "unmasked"]]

plt.figure(figsize=(6,4))

ax = plot_df.plot(kind="bar")

plt.ylabel("Number of significant bins")
plt.title("Site bias comparison: Masked vs Unmasked")
plt.xticks(rotation=0)
# ---- Add numeric labels ----
for container in ax.containers:
    for bar in container:
        height = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width()/2,
            height + max(plot_df.max())*0.005,
            f"{int(height)}",
            ha='center',
            va='bottom',
            fontsize=9,
            fontweight='bold'
        )

plt.tight_layout()

out_png = os.path.join(OUTDIR, "Site_bias_comparison_masked_vs_unmasked.png")
out_pdf = os.path.join(OUTDIR, "Site_bias_comparison_masked_vs_unmasked.pdf")

plt.savefig(out_png, dpi=300)
plt.savefig(out_pdf)

plt.show()

print("Saved:")
print(out_png)
print(out_pdf)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

MASKED = "/ix1/alee/LO_LAB/Personal/Rahul/cfDNA_project/Broad/NST/Fragmentation_analysis/Regional_Fragment_analysis/LOLA_inputs_top3sites/Effect_size/KW_BH/KW_BH.universe50.rawKWp_sig.with_site_direction.with_genes.tsv"
UNMASKED = "/ix1/alee/LO_LAB/Personal/Rahul/cfDNA_project/Broad/NST/Fragmentation_analysis/Regional_Fragment_analysis/LOLA_inputs_top3sites/Effect_size_no_cnv_masking/KW_BH/KW_BH.no_cnv_masking.rawsig.with_site_direction.with_genes.tsv"

OUTDIR = "/ix1/alee/LO_LAB/Personal/Rahul/cfDNA_project/Broad/NST/Fragmentation_analysis/Regional_Fragment_analysis/LOLA_inputs_top3sites/Comparison_CNV_noCNV/"
os.makedirs(OUTDIR, exist_ok=True)

df_mask = pd.read_csv(MASKED, sep="\t", dtype={"bin_id": str})
df_unmask = pd.read_csv(UNMASKED, sep="\t", dtype={"bin_id": str})

# Significant sets
A = set(df_mask["bin_id"].astype(str))
B = set(df_unmask["bin_id"].astype(str))

inter = len(A & B)
union = len(A | B)
jacc = inter / union if union else np.nan

only_A = len(A - B)
only_B = len(B - A)

print("Masked significant (A):", len(A))
print("Unmasked significant (B):", len(B))
print("Intersection (A∩B):", inter)
print("Union (A∪B):", union)
print(f"Jaccard = {jacc:.4f}")
print("Masked-only (A\\B):", only_A)
print("Unmasked-only (B\\A):", only_B)

# =========================
# Plot
# =========================

plt.figure(figsize=(6,5))

bars = ["Intersection", "Masked-only", "Unmasked-only"]
values = [inter, only_A, only_B]

bars_plot = plt.bar(bars, values)

plt.title(f"Significant bin overlap\nJaccard = {jacc:.3f}")
plt.ylabel("Number of bins")

# ---- Add simple n= labels above bars ----
for i, v in enumerate(values):
    plt.text(
        i,
        v + max(values)*0.0,   # small vertical offset
        f"n={v}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

plt.tight_layout()

out_png = os.path.join(OUTDIR, "Jaccard_overlap_breakdown.png")
out_pdf = os.path.join(OUTDIR, "Jaccard_overlap_breakdown.pdf")

plt.savefig(out_png, dpi=300)
plt.savefig(out_pdf)
plt.show()
```
