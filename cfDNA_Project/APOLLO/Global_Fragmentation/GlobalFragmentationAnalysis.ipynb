```python
import os
import glob
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import io

# Define directory path
directory = "/bgfs/alee/LO_LAB/Personal/Rahul/cfDNA_project/Batch1_01242025/LPWGS_Kappa/GlobalFragmentationAnalysis/CollectInsertSizeMetrics"

# Define output directory
output_directory = directory

# Use glob to expand the wildcard pattern to match all relevant text files
file_paths = glob.glob(os.path.join(directory, "*.metrics.txt"))

# Extract sample names from TXT file names
samplenames = [os.path.basename(x).replace(".metrics.txt", "") for x in file_paths]

# Create an empty DataFrame to hold all samples with insert_size as columns
combined_data = pd.DataFrame()

# Dictionary to hold the proportion of fragments below 150 bp for each sample
proportion_below_150 = {}

# Loop through each file, calculate density, and add to the DataFrame
for file_path, sample_name in zip(file_paths, samplenames):
    print(f"Processing: {file_path}")
    
    # Read file lines
    with open(file_path, 'r') as file:
        lines = file.readlines()
    
    # Find start indices for sections
    histogram_start = next((i for i, line in enumerate(lines) if line.startswith("## HISTOGRAM")), None)
    
    if histogram_start is None:
        print(f"Warning: No histogram section found in {file_path}")
        continue
    
    # Extract histogram section
    histogram_section = lines[histogram_start + 1 :]
    
    # Parse histogram data
    try:
        histogram_df = pd.read_csv(io.StringIO("\n".join(histogram_section)), sep="\t", comment='#')
    except pd.errors.EmptyDataError:
        print(f"Warning: No histogram data found in {file_path}")
        continue
    
    # Ensure histogram data is not empty
    if histogram_df.empty or 'insert_size' not in histogram_df or 'All_Reads.fr_count' not in histogram_df:
        print(f"Warning: Missing required columns in histogram data for {file_path}")
        continue
    
    # Convert columns to numeric if needed
    histogram_df['insert_size'] = pd.to_numeric(histogram_df['insert_size'], errors='coerce')
    histogram_df['All_Reads.fr_count'] = pd.to_numeric(histogram_df['All_Reads.fr_count'], errors='coerce')
    
    # Drop rows with NaN values in required columns
    histogram_df = histogram_df.dropna(subset=['insert_size', 'All_Reads.fr_count'])
    
    # Filter out insert sizes below 90
    histogram_df = histogram_df[histogram_df['insert_size'] >= 90]
    
    # Calculate density as a percentage
    histogram_df['density'] = (histogram_df['All_Reads.fr_count'] / histogram_df['All_Reads.fr_count'].sum()) * 100
    
    # Calculate the proportion of fragments below 150 bp
    total_counts = histogram_df['All_Reads.fr_count'].sum()
    below_150_counts = histogram_df[histogram_df['insert_size'] < 150]['All_Reads.fr_count'].sum()
    proportion = below_150_counts / total_counts
    proportion_below_150[sample_name] = proportion
    
    # Pivot the DataFrame so insert_size becomes columns and density becomes the values
    histogram_df = histogram_df.pivot_table(index=None, columns='insert_size', values='density')
    histogram_df.index = [sample_name]  # Rename the index to match the sample name
    
    # Append the data to the combined DataFrame
    combined_data = pd.concat([combined_data, histogram_df])

# Save the proportion of fragments below 150 bp to a CSV file
proportion_df = pd.DataFrame(list(proportion_below_150.items()), columns=['Sample', 'Proportion_Below_150'])
proportion_df.to_csv(os.path.join(output_directory, "proportion_below_150.csv"), index=False)

# Ensure the insert_size columns are sorted in increasing order
combined_data = combined_data.reindex(sorted(combined_data.columns), axis=1)

# Fill NaN values with 0, if any exist
combined_data = combined_data.fillna(0)

# Plotting the heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(combined_data, cmap='viridis', linewidths=0.0)
plt.title('Density Heatmap of Insert Sizes for Multiple Samples')
plt.xlabel('Fragment Size (bp)')
plt.ylabel('Samples')

# Determine X-axis range dynamically
max_insert_size = max(combined_data.columns) if not combined_data.empty else 299
plt.xlim(90, max_insert_size)

# Dynamically filter valid x-ticks
desired_ticks = [90, 100, 150, 167, 200, 250, 299]
available_ticks = [x for x in desired_ticks if x in combined_data.columns]
tick_positions = [combined_data.columns.get_loc(x) for x in available_ticks]

plt.xticks(ticks=tick_positions, labels=available_ticks)

# Add vertical dashed line at 167
if 167 in combined_data.columns:
    plt.axvline(x=tick_positions[available_ticks.index(167)], linewidth=1, color="black", alpha=1, zorder=10, linestyle="--")

# Save the heatmap
plt.savefig(os.path.join(output_directory, "density_heatmap.png"), dpi=300)

# Display the heatmap
plt.show()

print("Density heatmap saved successfully.")
```
