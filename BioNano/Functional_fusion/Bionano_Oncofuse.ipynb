```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# File paths
file_path = '/ix1/alee/LO_LAB/Personal/Rahul/BioNano/Functional_Fusion/Oncofuse/Oncofuse_hg38_result.xlsx'
sheet_name = 'Oncofuse_hg38_result'
output_path = '/ix1/alee/LO_LAB/Personal/Rahul/BioNano/Functional_Fusion/Oncofuse/fusion_driver_probability_plot.png'

def plot_fusion_driver_probability(file_path, sheet_name, output_path):
    # Validate file path
    if not os.path.exists(file_path):
        print(f"Error: File not found at {file_path}")
        return

    try:
        # Load the specified sheet from the Excel file
        df = pd.read_excel(file_path, sheet_name=sheet_name)

        # Verify the 'DRIVER_PROB' column exists
        if 'DRIVER_PROB' not in df.columns:
            print("Error: 'DRIVER_PROB' column not found in the specified sheet.")
            return

        # Extract relevant data
        driver_probs = df['DRIVER_PROB'].dropna()

        # Prepare data for plotting
        plot_data = pd.DataFrame({
            'Fusion Index': range(1, len(driver_probs) + 1),
            'Driver Probability': driver_probs.values
        })

        # Plot
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=plot_data, x='Fusion Index', y='Driver Probability', s=20, alpha=0.7, color='grey')
        plt.xlabel('Number of Gene Transcript Fusions', fontsize=18)
        plt.ylabel('Fusion Driver Probability', fontsize=18)
        plt.grid(alpha=0.0)

        # Adjust tick label size
        plt.xticks(fontsize=16)
        plt.yticks(fontsize=16)
        
        # Thicken axis lines
        plt.gca().spines['bottom'].set_linewidth(2)
        plt.gca().spines['left'].set_linewidth(2)
        plt.gca().spines['right'].set_visible(False)
        plt.gca().spines['top'].set_visible(False)

        plt.tight_layout()
        plt.savefig(output_path, dpi=300)
        plt.show()

    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
plot_fusion_driver_probability(file_path, sheet_name, output_path)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# File paths
file_path = '/ix1/alee/LO_LAB/Personal/Rahul/BioNano/Functional_Fusion/Oncofuse/Oncofuse_hg38_result.xlsx'
sheet_name = 'Overlapping_gene'
output_path = '/ix1/alee/LO_LAB/Personal/Rahul/BioNano/Functional_Fusion/Oncofuse/Overlapping_gene_fusion_driver_probability_Gene_name_plot.png'

def plot_fusion_driver_probability(file_path, sheet_name, output_path):
    # Validate file path
    if not os.path.exists(file_path):
        print(f"Error: File not found at {file_path}")
        return
    
    try:
        # Load the specified sheet from the Excel file
        df = pd.read_excel(file_path, sheet_name=sheet_name)
        
        # Verify necessary columns
        required_columns = {'DRIVER_PROB', '5_FPG_GENE_NAME', '3_FPG_GENE_NAME'}
        if not required_columns.issubset(df.columns):
            print(f"Error: Required columns {required_columns} not found.")
            return

        # Extract relevant data
        driver_probs = df['DRIVER_PROB'].dropna()
        gene_fusions = df[['5_FPG_GENE_NAME', '3_FPG_GENE_NAME']].astype(str).agg('-'.join, axis=1)

        # Prepare data for plotting
        plot_data = pd.DataFrame({
            'Fusion Index': range(1, len(driver_probs) + 1),
            'Driver Probability': driver_probs.values,
            'Gene Fusion': gene_fusions.values
        })

        # Plot
        plt.figure(figsize=(14, 10))
        sns.scatterplot(data=plot_data, x='Fusion Index', y='Driver Probability', s=50, alpha=0.7, color='grey')
        plt.xlabel('Fusion Index', fontsize=14)
        plt.ylabel('Fusion Driver Probability', fontsize=14)
        plt.grid(alpha=0.0)

        # Adjust tick label size
        plt.xticks(fontsize=12)
        plt.yticks(fontsize=12)
        
        # Thickening the axis lines
        plt.gca().spines['bottom'].set_linewidth(2)
        plt.gca().spines['left'].set_linewidth(2)
        plt.gca().spines['right'].set_visible(False)
        plt.gca().spines['top'].set_visible(False)

        # Label points with probability > 0.9
        high_prob_data = plot_data[plot_data['Driver Probability'] > 0.9]
        for _, row in high_prob_data.iterrows():
            plt.text(row['Fusion Index'], row['Driver Probability'], row['Gene Fusion'], fontsize=10, ha='right', va='bottom')

        plt.tight_layout()
        plt.savefig(output_path, dpi=300)
        plt.show()
        print(f"Plot saved at: {output_path}")

    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
plot_fusion_driver_probability(file_path, sheet_name, output_path)
```
